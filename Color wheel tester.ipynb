```python
import os 
import imageio
import matplotlib.pyplot as plt
from matplotlib.pyplot import imshow as imshow
```

```python
folder = input("Folder")
print(folder)
```

# Folder finding

```python
cwd = os.getcwd()
print(cwd)
```

```python
# Function to find all jpgs in the folder
image_list = []
def ourpath(path,folder):
    folderpath = path+'/'+folder
    for name in os.listdir(folderpath):
        newpath = folderpath+'/'+name
        if os.path.isfile(newpath) and newpath.endswith(".jpg"):
            print(newpath)
            image_list.append((newpath,name))
    print(len(image_list))

ourpath(os.getcwd(), folder)
```

```python
for image in image_list:
    print(image[1])
```

# Image Creation

```python
rgb_name = [(0,"Red"),(1,"Green"),(2,"Blue")]
```

```python
# Create pngs - 4 versions for each original image. Red, Green, Blue, and Black/white
plt.rcParams['axes.titlesize'] = 100
for image in image_list:
    plt.rcParams['figure.facecolor'] = "#7F7F7F" #Grey background for sensitive eyes
    for i in rgb_name:
        iamge = imageio.v2.imread(str(image[0]))
        size = iamge.shape[0], iamge.shape[1]
        fig_width = size[0] / 100
        fig_height = size[1] / 100
        fig, ax = plt.subplots(figsize=(fig_width, fig_height), dpi=100)

        ax.set_title(i[1])
        ax.imshow(iamge[:,:,i[0]],cmap="gray", vmin=0., vmax=255)
        ax.axes.get_xaxis().set_visible(False)
        ax.axes.get_yaxis().set_visible(False)
        filename = str(image[1])[:-4]+"_"+str(i[0])+".png"
        print(filename)
        fig.savefig(filename, pad_inches=0)
        fig.clf()
        plt.close()
        ax.cla()
    iamge = imageio.v2.imread(str(image[0]))
    size = iamge.shape[0], iamge.shape[1]
    fig_width = size[0] / 100
    fig_height = size[1] / 100
    fig, ax = plt.subplots(figsize=(fig_width, fig_height), dpi=100)

    image_grey = iamge.mean(-1)/iamge.max()
    ax.set_title("Black/white")
    ax.imshow(image_grey,cmap="gray", vmin=0., vmax=1)
    ax.axes.get_xaxis().set_visible(False)
    ax.axes.get_yaxis().set_visible(False)
    filename = str(image[1])[:-4]+"_3"+".png"
    print(filename)
    fig.savefig(filename, pad_inches=0)
    fig.clf()
    plt.close()
    ax.cla()
```

# How many of each color

```python
# Function to find and order jpgs in the folder by color
def lastitem(word_freq):
    return word_freq[1]

colordic = {}
colors = ["Red","Orange","Yellow","Lime-green","Green","Seafoam","Cyan","Blue","Purple","Pink","Grey","Light-grey","Brown","Teal","Bordeaux","Sky-blue","Lavender","Blonde"]
def colorfinder(path,folder):
    folderpath = path+'/'+folder
    for name in os.listdir(folderpath):
        newpath = folderpath+'/'+name
        if os.path.isfile(newpath) and newpath.endswith(".jpg"):
            for color in colors:
                if color in newpath:
                    if color in colordic: colordic[color] += 1
                    else: colordic[color] = 1
    print(sorted(colordic.items(), key=lastitem, reverse=False))
colorfinder(os.getcwd(), folder)
```
